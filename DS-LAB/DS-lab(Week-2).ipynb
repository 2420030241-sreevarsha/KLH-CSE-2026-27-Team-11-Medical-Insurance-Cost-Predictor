```python
import pandas as pd
import numpy as np
#sample dataset
df=pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
                 'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']})
print(df)
#mean for numeric
df['Age']=df['Age'].fillna(df['Age'].mean())
#mode for categorical
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
#forward fill(use previous value)
import pandas as pd
import numpy as np
#sample dataset
df=pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
                 'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']})
#display original dataset
print("original dataSet(with missing value):")
print(df)
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
```

```python
#backward fill(use after value)
import pandas as pd
import numpy as np
#sample dataset
df=pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
                 'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']})
#display original dataset
print("original dataSet(with missing value):")
print(df)
df_bfill=df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
#drop rows with missing values
import pandas as pd
import numpy as np
#sample dataset
df=pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
                 'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']})
print('Orginal dataSet')
print(df)
df_drop_rows=df.dropna()
#drop rows with missing values
print('After dropping')
print(df_drop_rows) #deletes the rows which contains missing values
```

```python
#drop columns with missing values
df_col_drop=df.dropna(axis=1)
print("After dropping columns:\n",df_col_drop)
```

```python
#removal of the duplicates
import pandas as pd
df=pd.DataFrame({
    'ID':[1, 2, 2, 3, 4, 4],
    'Name':['Alice', 'bob', 'bob', 'charlie', 'david', 'david'],
    'Age':['25', '30', '30', '35', '40', '40']
})
print("original data:\n",df)
df_exact=df.drop_duplicates()
print("After exact match removal:\n",df_exact)
```

```python
#removing a subset
#remove duplicates based on
import pandas as pd
df=pd.DataFrame({
    'ID':[1, 2, 2, 3, 4, 4],
    'Name':['Alice', 'bob', 'bob', 'charlie', 'david', 'david'],
    'Age':['25', '30', '30', '35', '40', '40']
})
print("original data:\n",df)

#remove duplicates based only on 'ID'
df_subset_id=df.drop_duplicates(subset=['ID'])
print("After subset based removal(ID):\n",df_subset_id)

#remove duplicates based only on 'Name'
df_subset_Name=df.drop_duplicates(subset=['Name'])
print("After subset based removal(Name):\n",df_subset_Name)

```

```python
import pandas as pd
df=pd.DataFrame({
    'Date':['2025-01-05', '05/01/2025', 'Jan 5 2025', '2025.01.05']
})
print("original data:\n",df)
df['Date']=pd.to_datetime(df['Date'], errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```
